```
%%writefile firsttry.cpp
#include <iostream>
#include <fstream>
#include <string>
#include <sstream>
#include <iomanip>
#include <omp.h>
#include <vector>
#include <thread>

bool e_numero(const std::string& s) {
    if (s.empty()) return false;

    int numero = std::atoi(s.c_str());


    if (numero != 0) {
        return true;
    }

    if (s[0] == '0') {
        return true;
    }

    return false;
}



void carregarArquivo(std::string caminho, std::vector<std::vector<std::string>>& dados) {
    std::ifstream arquivo(caminho);
    std::string linha;

    int largura = 17;

    if (arquivo.is_open()) {
        // Le o arquivo linha por linha
        while (std::getline(arquivo, linha)) {
            std::vector<std::string> colunas;
            std::stringstream ss(linha);
            std::string palavra;

            // Quebra a linha em palavras (baseado em espaços ou abas)
            while (ss >> palavra) {
                std::cout << std::left << std::setw(largura) << palavra;
                std::string tipo = "";
                if (e_numero(palavra) == true) {
                    tipo = "´[NUM]";
                } else {
                    tipo = "[STR]";
                }
                std::cout << std::left << std::setw(largura) << (tipo + palavra);
                colunas.push_back(palavra);
            }

            // Se a linha não estiver vazia, adiciona ao vetor principal
            if (!colunas.empty()) {
                std::cout << "|" << "\n";
                dados.push_back(colunas);
            }
        }
        arquivo.close();
        std::cout << "Thread - Leitura de matriz finalizada com sucesso.\n";
    } else {
        std::cerr << "Thread - Erro ao abrir o arquivo.\n";
    }
}


void f_numero(){ //funcao para estatistica descritiva
  std::cout << "sou a thread 1!" << std::endl;
}

void f_string(int _par1){ //funcao para os dicionarios
  std::cout << "sou a thread 2 com o parametro: " << _par1 << "!" << std::endl;
}


int main() {
  std::cout << "criando e executando as threads..." << std::endl; //"endl" pula linha tipo o /n
  std::vector<std::vector<std::string> > vdados;
  std::string Arquivo = "/content/Data_SET_Piva - Data_SET_Piva.csv";



  std::thread worker(carregarArquivo, Arquivo, std::ref(vdados));


  std::thread th1(f_numero);
  std::thread th2(f_string, 2026);
  worker.join();
  th1.join();
  th2.join();
  std::cout << "Total de elementos lidos: " << vdados.size() << "\n";
return 0;
}
```
